```python
import numpy as np
import matplotlib.pyplot as plt

# Muestras
N = 8

n = np.arange(0, N, 1) # vector de variable discreta

# Defino función 
def x(n):
    return 4 + 3*np.sin(np.pi/2*n)

# FUNCION
x1 = x(n) 

# DFT
x2 = np.fft.fft(x1) #DFT

# MODULO Y FASE
xabs = np.abs(x2)**2 *(1/N**2)
xfase = np.angle(x2)

# GRAFICO
plt.figure(figsize=(10, 8))

# --- x[n] ---
plt.subplot(3, 1, 1)
plt.stem(n, x1, basefmt=" ")
plt.axhline(0, color='black', linewidth=1)
plt.title('Señal discreta x[n]')
plt.xlabel('n')
plt.ylabel('Amplitud')
plt.grid(True)

# --- |X[k]| ---
plt.subplot(3, 1, 2)
plt.stem(n, xabs, basefmt=" ")
plt.axhline(0, color='black', linewidth=1)
plt.title('Módulo |X[k]|')
plt.xlabel('k')
plt.ylabel('Magnitud')
plt.grid(True)

# --- Fase ---
plt.subplot(3, 1, 3)
plt.stem(n, xfase, basefmt=" ")
plt.axhline(0, color='black', linewidth=1)
plt.title('Fase ∠X[k]')
plt.xlabel('k')
plt.ylabel('Fase [rad]')
plt.grid(True)

plt.tight_layout()
plt.show()
```

La energía de la señal se concentra en los índices k=0,2,6, ya que son los únicos valores donde los coeficientes de la DFT son distintos de cero.
El término en k=0 corresponde al valor medio de la señal, mientras que k=2 y k=6 corresponden a la frecuencia de la senoide.
